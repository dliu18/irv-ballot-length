```python
import numpy as np
import matplotlib.pyplot as plt

import sys, os
sys.path.append(os.path.abspath(os.path.join(os.getcwd(), '..')))

import utils
from irv import run_irv
```

```python
burlington_filename = "../data/preflib/elections-all/burlington/ED-00005-00000002.toi"

ballots, ballot_counts, cand_names, skipped_votes = \
    utils.read_preflib(burlington_filename)

elim_votes = run_irv(
    len(cand_names),
    ballots.copy(), 
    ballot_counts, 
    cands=cand_names)

elim_order = utils.get_elim_order(elim_votes).split(" ")
print(elim_votes)
print(elim_order)
```

```python
sampling_rate = 0.01
h = 4
num_trials = 100
num_matches = 0

for trial_num in range(num_trials):
    total_n = int(np.sum(ballot_counts))
    poll_size = max(int(total_n * sampling_rate), 1)
    
    # Draw a poll from the full election
    poll = utils.resample(
        ballot_counts,
        sample_size=poll_size,
        with_replacement=False,
        seed=trial_num,
    )
    
    sampler = "Bootstrap"
    sampling_model = utils.get_model_object(sampler, cand_names)
    sampling_model.fit(ballots, poll)
    
    simulated_ballots, simulated_counts = sampling_model.simulate_ballots(
        num_ballots = total_n
    )
    
    simulated_elim_votes = run_irv(
        len(cand_names),
        simulated_ballots.copy(), 
        simulated_counts, 
        cands=cand_names)
    
    simulated_elim_order = utils.get_elim_order(simulated_elim_votes).split(" ")
    
    overlap = set(simulated_elim_order[-h:]).intersection(
        set(elim_order[-h:])
    )

    # if simulated_elim_order[-1] == "1":
    #     num_matches += 1
    if len(overlap) == h:
        num_matches += 1    
    # if simulated_elim_order[-h:] != elim_order[-h:]:
    #     print(simulated_elim_order)

print(num_matches)
```

```python
filtered_cands = [1, 2, 3, 4, 5, 6, 7]
filtered_cand_names = {cand: full_name for cand, full_name in cand_names.items() if cand in filtered_cands}
filtered_ballots, filtered_ballot_counts = utils.reduce_election(ballots, ballot_counts, filtered_cands)

sampling_rate = 0.1
num_trials = 100
num_matches = 0

for trial_num in range(num_trials):
    total_n = int(np.sum(ballot_counts))
    poll_size = max(int(total_n * sampling_rate), 1)
    
    # Draw a poll from the full election
    poll = utils.resample(
        filtered_ballot_counts,
        sample_size=poll_size,
        with_replacement=False,
        seed=trial_num,
    )
    
    sampler = "Bootstrap"
    sampling_model = utils.get_model_object(sampler, filtered_cand_names)
    sampling_model.fit(filtered_ballots, poll)
    
    simulated_ballots, simulated_counts = sampling_model.simulate_ballots(
        num_ballots = total_n
    )
    
    simulated_elim_votes = run_irv(
        len(cand_names),
        simulated_ballots.copy(), 
        simulated_counts, 
        cands=cand_names)
    
    simulated_elim_order = utils.get_elim_order(simulated_elim_votes).split(" ")

    if simulated_elim_order[-1] == "4":
        num_matches += 1
    if simulated_elim_order[-h:] != elim_order[-h:]:
        print(simulated_elim_order)
print(num_matches)
```
